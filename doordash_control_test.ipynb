# Week 7 Bonus: Testing the controls on a toy DoorDash support chatbot

## The knowledge base
Policy articles the chatbot must use.

```python
kb = {
    "late_order": "If the restaurant is late, wait up to 10 minutes then contact support.",
    "wrong_address": "If the address is wrong, message the customer through the app.",
    "pay_issue": "Pay questions are answered in the Earnings tab of the Dasher app.",
    "order_issue": "If the order is missing items, take a photo and report it in the app.",
    "restaurant_issue": "If the restaurant is not open, wait 5 minutes then unassign the order.",
}

def words(text):
    return set(text.lower().replace(".", "").replace(",", "").replace("?", "").split())
```

## Control 1: Retrieval
Pick the best-matching article for a question.

```python
def retrieve(question):
    # Pick the article sharing the most words with the question
    return max(kb, key=lambda k: len(words(question) & words(kb[k])))
```

## The chatbot
Answers with the article it retrieved.

```python
def chatbot(question):
    article = retrieve(question)
    print("Driver asks:  ", question)
    print("Bot uses:     ", article)
    print("Bot replies:  ", kb[article])
    print()
```

## Control 2: Grounding check
Check whether an answer matches its article.

```python
def grounded(answer, article, threshold=0.5):
    # Share of the answer's words that also appear in the article
    a = words(answer)
    return len(a & words(kb[article])) / len(a) >= threshold
```

## Step 2: Ask questions as a driver

```python
print("=== Retrieval test ===\n")
for q in ["My order is missing a drink, what do I do?",
          "When do I get paid?"]:
    chatbot(q)
```

## Step 3: Test answers as the AI

```python
print("=== Answer check test ===\n")
tests = [
    # (answer to test,                     article,       should the control pass it?)
    ("Take a photo and report it in the app", "order_issue", True),
    ("Do not report it in the app",           "order_issue", False),
]

for i, (answer, article, should_pass) in enumerate(tests, 1):
    did_pass = grounded(answer, article)
    result = "Pass" if did_pass == should_pass else "EXCEPTION"
    print(i, answer, "| should pass:", should_pass, "| did pass:", did_pass, "|", result)
```
